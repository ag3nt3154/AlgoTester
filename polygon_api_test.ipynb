```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from polygon import RESTClient
from config.api_keys import POLYGON_API_KEY
import datetime

client = RESTClient(api_key=POLYGON_API_KEY)
ticker = 'SPY'
start_date = '1980-01-01'
end_date = '2025-04-10'
try:
    # Get the aggregates (OHLC) data
    aggs = []
    for a in client.list_aggs(
        ticker=ticker,
        multiplier=1,
        timespan="day",
        from_=start_date,
        to=end_date,
        limit=50000  # Max allowed by API
    ):
        aggs.append(a)
    
    if not aggs:
        print(f"No data found for {ticker} between {start_date} and {end_date}")
        
    
    # Convert to pandas DataFrame
    df = pd.DataFrame([{
        "date": datetime.datetime.utcfromtimestamp(agg.timestamp / 1000).date(),
        "open": agg.open,
        "high": agg.high,
        "low": agg.low,
        "close": agg.close,
        "volume": agg.volume,
        "vwap": agg.vwap,
        "transactions": agg.transactions
    } for agg in aggs])
    
    # Sort by date (ascending)
    df = df.sort_values("date").reset_index(drop=True)
    display(df)
except Exception as e:
    print(f"Error fetching data: {e}")
    
```
