# Load libraries and sample data

```python
import pandas as pd
import numpy as np
from utils.backtester import BackTester
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt



ticker_list = [
    "SPY",
    "GLD", 
    "TLT",
    "SHV",
]
price_data = {}
for ticker in ticker_list:
    # Load sample data
    df = pd.read_pickle(f'data/raw/price/{ticker}_20250206.pkl')
    df.set_index('Date', inplace=True)
    # data.index = pd.to_datetime(data.index, format="mixed")
    price_data[ticker] = df
```

# Trailing return trend following strategy

```python
from strategies.trend_following import TrendFollowingStrategy

def backtest(ticker, lookback_period=252):
    tl = [ticker]
    pda = {ticker: price_data[ticker]}
    # Initialize backtester
    bt = BackTester(pda)
    bt.add_strategy(TrendFollowingStrategy, tickers=tl, price_data=pda, lookback=lookback_period)
    results = bt.backtest()
    print("Trend following Results:", {k: v for k, v in results.items() if k != 'returns'})
    bt.plot_results()
    return

for ticker in ticker_list:
    if ticker == 'SHV': continue
    print("Backtesting ticker:", ticker)
    backtest(ticker)
```

## Optimize for lookback period

```python
lookback_arr = [5, 10, 20, 50, 100, 200, 252, 300, 400, 500]

for ticker in tqdm(ticker_list):
    if ticker == 'SHV': continue
    tl = [ticker]
    pda = {ticker: price_data[ticker]}
    sharpe_arr = []
    
    for lookback in tqdm(lookback_arr):
        bt = BackTester(pda)
        bt.add_strategy(TrendFollowingStrategy, tickers=tl, price_data=pda, lookback=lookback)
        results = bt.backtest()
        sharpe_arr.append(results['sharpe'])
    
    plt.plot(lookback_arr, sharpe_arr)
    plt.ylabel('Sharpe Ratio')
    plt.xlabel('Lookback Period')
    plt.title(ticker)
    plt.show()

    optimal_lookback = lookback_arr[sharpe_arr.index(max(sharpe_arr))]
    print(f'Optimal Lookback Period for {ticker}: {optimal_lookback}')

    bt = BackTester(pda)
    bt.add_strategy(TrendFollowingStrategy, tickers=tl, price_data=pda, lookback=optimal_lookback)
    results = bt.backtest()
    bt.plot_results()
```

## Dual lookback

Based on the paper: https://people.duke.edu/~charvey/Research/Published_Papers/P158_Momentum_turning_points.pdf

```python
from strategies.trend_following import DualLookbackTrendFollowing

weight_arr = np.linspace(0, 1, 10)

for ticker in tqdm(ticker_list):
    # filter out money market fund
    if ticker == 'SHV': continue

    tl = [ticker]
    pda = {ticker: price_data[ticker]}

    # optimize for rebound fast weight
    sharpe_arr = []
    for weight in tqdm(weight_arr):
        bt = BackTester(pda)
        bt.add_strategy(DualLookbackTrendFollowing, tickers=tl, price_data=pda, fast_lookback=21, slow_lookback=252, rebound_fast_weight=weight, correction_fast_weight=0.5)
        results = bt.backtest()
        sharpe_arr.append(results['sharpe'])
    plt.plot(weight_arr, sharpe_arr)
    plt.ylabel('Sharpe Ratio')
    plt.xlabel('Lookback weight')
    plt.title(ticker)
    plt.show()
    optimal_rebound_weight = weight_arr[sharpe_arr.index(max(sharpe_arr))]

    # optimize for correction fast weight
    sharpe_arr = []
    for weight in tqdm(weight_arr):
        bt = BackTester(pda)
        bt.add_strategy(DualLookbackTrendFollowing, tickers=tl, price_data=pda, fast_lookback=21, slow_lookback=252, correction_fast_weight=weight, rebound_fast_weight=0.5)
        results = bt.backtest()
        sharpe_arr.append(results['sharpe'])
    plt.plot(weight_arr, sharpe_arr)
    plt.ylabel('Sharpe Ratio')
    plt.xlabel('Lookback weight')
    plt.title(ticker)
    plt.show()
    optimal_correction_weight = weight_arr[sharpe_arr.index(max(sharpe_arr))]


    print(f'Optimal rebound fast weight for {ticker}: {optimal_rebound_weight}')
    print(f'Optimal correction fast weight for {ticker}: {optimal_correction_weight}')

    bt = BackTester(pda)
    bt.add_strategy(DualLookbackTrendFollowing, tickers=tl, price_data=pda, fast_lookback=21, slow_lookback=252, correction_fast_weight=optimal_correction_weight, rebound_fast_weight=optimal_rebound_weight)
    results = bt.backtest()
    print("Trend following Results:", {k: v for k, v in results.items() if k != 'returns'})
    bt.plot_results()
```

# Moving average crossover

```python
from strategies.trend_following import MovingAverageCrossoverStrategy

def backtest_sma(ticker, short_window, long_window, rebalance_period):
    print("Backtesting SMA strategy for {}".format(ticker))
    tl = [ticker]
    pda = {ticker: price_data[ticker]}

    bt = BackTester(pda)
    bt.add_strategy(MovingAverageCrossoverStrategy, tickers=tl, price_data=pda, short_window=short_window, long_window=long_window, rebalance_period=rebalance_period)
    results = bt.backtest()
    print("Trend following Results:", {k: v for k, v in results.items() if k != 'returns'})
    bt.plot_results()

for ticker in ticker_list:
    if ticker == 'SHV': continue
    backtest_sma(ticker, 21, 252, 21)
```
